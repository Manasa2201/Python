Type casting

```python
# integer typecasting : int()
v1 = int(5.6)  # float to int
v2 = int("500") # string to int
v3 = int(True) # Bool to Int
v4 = int(False) # Bool to Int
v5 = int("1010", 2)

print(v1, type(v1))
print(v2, type(v2))
print(v3, type(v3))
print(v4, type(v4))
print(v5, type(v5))
```

```python
# Convert to float: float()
v1 = float(105)            # int to float
v2 = float("4.5")
v3 = float(True)

print(v1, type(v1))
print(v2, type(v2))
print(v3, type(v3))
```

```python
# convert to string: str()
v1 = str(200)
v2 = str(4.5)
v3 = str(True)

print(v1, type(v1))
print(v2, type(v2))
print(v3, type(v3))
```

```python
# Convert to Boolean: bool()
v1 = bool(1)
v2 = bool(0)
v3 = bool(1000)
v4 = bool("Hello")

print(v1, type(v1))
print(v2, type(v2))
print(v3, type(v3))
print(v4, type(v4))
```
