1. Arithmatic operators (+, -, *, /, %, //, **)

```python
a = 10
b = 3

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)
print(a%b)
print(a**b)
```

```python
a = -10
b = 3

print(a//b)
```

2. Bitwise operators (<<, >>, ~, &, |, ^)

```python
# Bitwise operators works on binary number
# Bitwise operators are faster than arithmatic operators
# Bitwise operators perform many operation in easy way due to binary number system.
```

```python
a = 10
b = 3

print(a & b)
print(a | b)
print(a ^ b)
```

```python
a = 10
b = 3
c = 2
print(a & b & c)
```

```python
print(a | b | c)
```

```python
print(a ^ b ^ c)
```

left shift (<<) , Right shift (>>)

```python
# Def (left shift): Insert new bits (0) to the LSB
# Def (Right shift) : Delete the bits from the LSB
```

```python
a = 10
print(a << 1)  # left shift by 1 bit
```

```python
print(a << 2)  # left shift by 2 bit
```

```python
print(a << 3)  # left shift by 3 bit
```

```python
print(a << 6)  # left shift by 8 bit
```

```python
# Right shift (>>)

a = 10
print(a >> 1)
```

```python
print(a >> 2)
```

```python
print(a >> 3)
```

bitwise not (~)

```python
# def: flip the bits
a = 12
print(~a)
```

```python
a = 10
print(~a)
```

```python
# Negative number
a = -100
print(~a)
```

3. Assignment operators (+=, -=, *=, /=,...)

```python
a = 10

a+=5  # a = a + 5
print(a)

a-=5  # a = a - 5
print(a)

a*=5  # a = a * 5
print(a)

a/=5  # a = a / 5
print(int(a))

a%=5  # a = a * 5
print(a)

# &=, |=, ^=, <<=, >>=, :=

a = 10
a &= 3
print(a)
```

```python
# walrus operator (:=)
# walrus operator will first assign the value to the variable then pass it to outer function

print(y := 2)
```

4. Comparison operators (==, >=, <=, !=, <, >)

```python
a = 10
b = 5

print(a == b)
print(a >= b)
print(a <= b)
print(a != b)
print(a < b)
```

Logical operators

```python
# and, or, not
a = True
b = False

print(a and b)
print(a or b)
print(not b)
```

identity operators (is, is not)

```python
a = [1,2,3]
b = [1,2,3]
c = a

print(a == b)

print(a is b)

print(a is c)

print(a is not c)
```

Membership operators

```python
# in, not in

fruits = ['Mamidipandu', 'AratiPandu', 'Batai', 'Panasapandu']

print('Batai' in fruits)

print('Batai' not in fruits)
```
